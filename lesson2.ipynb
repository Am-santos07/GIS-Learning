```python
from shapely.geometry import Point, LineString

def create_line_geometry(point_list):

    # Check that the input is a list
    if not isinstance(point_list, list):
        raise TypeError("Input should be a list")

    # Check that the list contains at least two points
    assert len(point_list) >= 2, "At minimum two points are required for a LineString"

    # Check that all values are Shapely Point objects
    assert all(isinstance(point, Point) for point in point_list), "All list values must be of type shapely.geometry.Point"

    # Create and return the LineString
    line = LineString(point_list)
    return line
```

```python
# Create two Point objects
point1 = Point(45.2, 22.34)
point2 = Point(100.22, -3.20)

# Create a LineString
line1 = create_line_geometry([point1, point2])

# Display the result
print(line1)
print(line1.geom_type)
```

```python
from shapely.geometry import Point, Polygon

def create_polygon_geometry(coordinates):

    # Check that the input is a list
    assert isinstance(coordinates, list), "Input should be a list"

    # Check that the list contains at least three points
    assert len(coordinates) >= 3, "At minimum three points are required for a polygon"

    # Check if all values are coordinate tuples
    if all(isinstance(coord, tuple) for coord in coordinates):

        # Check that every tuple contains exactly two values
        assert all(len(coord) == 2 for coord in coordinates), "All list values must be coordinate tuples"

        # Check that coordinate values are numbers
        assert all(
            isinstance(value, (int, float))
            for coord in coordinates
            for value in coord
        ), "Coordinate values must be numbers"

        return Polygon(coordinates)

    # Allow a list of Shapely Point objects
    elif all(isinstance(coord, Point) for coord in coordinates):
        return Polygon(coordinates)

    else:
        raise AssertionError("All list values must be coordinate tuples or Shapely Points")
```

```python
# Create a polygon using three coordinate tuples

coordinates = [
    (45.2, 22.34),
    (100.22, -3.20),
    (70.0, 10.20)
]

# Create the polygon
polygon1 = create_polygon_geometry(coordinates)
```

```python
print(polygon1)
print(polygon1.geom_type)
```

```python
from shapely.geometry.base import BaseGeometry

def get_centroid(geom):

    # Check that the input is a Shapely geometry
    assert isinstance(geom, BaseGeometry), "Input must be a `shapely` geometry"

    # Calculate and return the centroid
    return geom.centroid
```

```python
# Calculate the centroid of polygon1
centroid = get_centroid(polygon1)

# Display the centroid
print(centroid)
print(centroid.geom_type)
```

```python
try:
    # Pass something else than a Shapely geometry
    get_centroid("Give me a centroid!")

except AssertionError as exception:
    print(f"The function (correctly) detected an error. The error message was '{exception.args[0]}'")

except Exception as exception:
    raise exception
```

```python
from shapely.geometry import Polygon

def get_area(polygon):

    # Check that the input is a Shapely Polygon
    assert isinstance(polygon, Polygon), "Input should be a `shapely.geometry.Polygon`"

    # Calculate and return the polygon area
    return polygon.area
```

```python
# Calculate the area
area = get_area(polygon1)

# Display the area rounded to 2 decimal places
print(round(area, 2))
```

```python
from shapely.geometry import LineString, Polygon

def get_length(geometry):

    # Check whether the geometry is a LineString
    if isinstance(geometry, LineString):
        return geometry.length

    # Check whether the geometry is a Polygon
    elif isinstance(geometry, Polygon):
        return geometry.exterior.length

    # Raise an error for any other geometry type
    else:
        raise ValueError("'geometry' should be either a LineString or a Polygon")
```

```python
# Calculate the length of the LineString
line_length = get_length(line1)

print("Line length:", round(line_length, 2))


# Calculate the exterior length of the Polygon
poly_exterior_length = get_length(polygon1)

print("Polygon exterior length:", round(poly_exterior_length, 2))
```

```python
try:
    # Pass something else than a Shapely LineString or Polygon
    get_length(Point(1, 2))

except (AssertionError, ValueError) as exception:
    print(f"The function (correctly) detected an error. The error message was '{exception.args[0]}'")

except Exception as exception:
    raise exception
```
